```python
%%bash

echo "=== OpenSSL Version ==="
openssl version
```

```python
%%bash

openssl genrsa -out sign_private.pem 2048

echo "=== RSA-2048 Private Key Created ==="
echo "File: sign_private.pem"
```

```python
%%bash

openssl rsa \
  -in sign_private.pem \
  -pubout \
  -out sign_public.pem

echo "=== RSA-2048 Public Key Created ==="
echo "File: sign_public.pem"
```

```python
%%bash

echo "Hello! This is my RSA digital signature." > message.txt

echo "=== Message ==="
cat message.txt
```

```python
%%bash

openssl dgst \
  -sha256 \
  -sign sign_private.pem \
  -out signature.bin \
  message.txt

echo "=== Digital Signature Created ==="
echo "Signature file: signature.bin"
```

```python
%%bash

echo "=== Digital Signature (Base64) ==="

openssl base64 \
  -in signature.bin
```

```python
%%bash

echo "=== Signature Verification ==="

openssl dgst \
  -sha256 \
  -verify sign_public.pem \
  -signature signature.bin \
  message.txt
```
