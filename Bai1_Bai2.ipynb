```python
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives import padding


with open("crypto_lab/aes/aes_key.bin", "rb") as f:
    key = f.read()

with open("crypto_lab/aes/iv.bin", "rb") as f:
    iv = f.read()


with open("crypto_lab/aes/ciphertext.bin", "rb") as f:
    ciphertext = f.read()

print("AES key:", len(key) * 8, "bits")
print("IV:", len(iv), "bytes")
print("Ciphertext:", len(ciphertext), "bytes")
```

```python
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives import padding

cipher = Cipher(
    algorithms.AES(key),
    modes.CBC(iv)
)

decryptor = cipher.decryptor()

padded_plaintext = (
    decryptor.update(ciphertext)
    + decryptor.finalize()
)

# AES-CBC của OpenSSL dùng PKCS#7 padding
unpadder = padding.PKCS7(128).unpadder()

plaintext = (
    unpadder.update(padded_plaintext)
    + unpadder.finalize()
)

print(plaintext.decode("utf-8"))
```

```python
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.primitives.asymmetric import padding
from cryptography.hazmat.primitives import hashes

# Đọc RSA private key do OpenSSL tạo
with open("crypto_lab/rsa/private.pem", "rb") as f:
    private_key_data = f.read()

private_key = serialization.load_pem_private_key(
    private_key_data,
    password=None
)

# Đọc encrypted AES key do OpenSSL tạo
with open("crypto_lab/rsa/encrypted_key.bin", "rb") as f:
    encrypted_key = f.read()

# RSA OAEP decrypt
decrypted_key = private_key.decrypt(
    encrypted_key,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)

print("Decrypted key length:", len(decrypted_key), "bytes")
```

```python
!openssl pkeyutl \
-encrypt \
-pubin \
-inkey crypto_lab/rsa/public.pem \
-in crypto_lab/rsa/aes_key.bin \
-out crypto_lab/rsa/encrypted_key.bin \
-pkeyopt rsa_padding_mode:oaep \
-pkeyopt rsa_oaep_md:sha256 \
-pkeyopt rsa_mgf1_md:sha256
```

```python
with open("crypto_lab/rsa/aes_key.bin", "rb") as f:
    original_key = f.read()

if original_key == decrypted_key:
    print("RSA decryption successful!")
else:
    print("RSA decryption failed!")
```

```python
from cryptography.hazmat.primitives import serialization

with open("crypto_lab/rsa/public.pem", "rb") as f:
    public_key_data = f.read()

public_key = serialization.load_pem_public_key(
    public_key_data
)
```

```python
with open("crypto_lab/signature/message.txt", "rb") as f:
    message = f.read()

with open("crypto_lab/signature/signature.bin", "rb") as f:
    signature = f.read()

print("Message:", message.decode())
print("Signature size:", len(signature), "bytes")
```

```python
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import padding

try:
    public_key.verify(
        signature,
        message,
        padding.PKCS1v15(),
        hashes.SHA256()
    )

    print("Signature verified successfully!")

except Exception:
    print("Signature verification failed!")
```

```python
with open("crypto_lab/signature/message_modified.txt", "rb") as f:
    modified_message = f.read()

try:
    public_key.verify(
        signature,
        modified_message,
        padding.PKCS1v15(),
        hashes.SHA256()
    )

    print("Signature verified successfully!")

except Exception:
    print("Signature verification failed!")
```
