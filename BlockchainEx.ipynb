```python
%%writefile plaintext.txt
Hello Blockchain!
This is my AES-256 encryption test.
```

```python
!openssl enc -aes-256-cbc -salt -pbkdf2 -a -in plaintext.txt -out encrypted.txt -pass pass:"MySecretPassword123"

!echo "Encrypted output:"
!cat encrypted.txt
```

```python
!openssl enc -d -aes-256-cbc -pbkdf2 -a -in encrypted.txt -out decrypted.txt -pass pass:"MySecretPassword123"

!echo "Decrypted output:"
!cat decrypted.txt
```
