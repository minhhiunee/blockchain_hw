```python
%%bash

openssl genrsa -out rsa_private.pem 2048

openssl rsa -in rsa_private.pem -pubout -out rsa_public.pem

echo "=== RSA Private Key ==="
openssl rsa -in rsa_private.pem -text -noout | head -15

echo ""
echo "=== RSA Public Key ==="
cat rsa_public.pem
```

```python
%%bash

echo "Hello RSA 2048!" > rsa_plaintext.txt

echo "=== Plaintext ==="
cat rsa_plaintext.txt
```

```python
%%bash

openssl pkeyutl -encrypt \
  -pubin \
  -inkey rsa_public.pem \
  -in rsa_plaintext.txt \
  -out rsa_encrypted.bin \
  -pkeyopt rsa_padding_mode:oaep \
  -pkeyopt rsa_oaep_md:sha256

echo "=== RSA Encrypted Output (Base64) ==="
openssl base64 -in rsa_encrypted.bin
```

```python
!openssl pkeyutl -decrypt \
  -inkey rsa_private.pem \
  -in rsa_encrypted.bin \
  -out rsa_decrypted.txt \
  -pkeyopt rsa_padding_mode:oaep \
  -pkeyopt rsa_oaep_md:sha256

!echo "=== RSA Decrypted Output ==="
!cat rsa_decrypted.txt
```
